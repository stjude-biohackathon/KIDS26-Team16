# 🏥 SCOGS — MedGemma 27B prompt ablation ladder (Colab, A100, Ollama)

Same machinery as `medgemma_27b_a100_16bit.ipynb`. What differs is Step 7: instead of one
run of one prompt, this runs the **ablation ladder** — the same 20 notes and the same four
outcomes through five cumulative prompt configurations — and Step 8 lays the rungs side by
side.

| rung | what it adds | judged on |
|:--|:--|:--|
| **0** | nothing — the prompt as it shipped | the control |
| **1** | JSON-schema decoding, field order, grouped rules, `repeat_penalty` 1.0, content retry | `null_placeholder` → ~0 |
| **2a** | the rubric's own Definition + Diagnostic Criteria, a quote behind `present` | `grade_set` → `graded` on 28/48/19 |
| **2b** | an `evidence` field, generated before the findings (chain-of-thought inside the JSON) | `proposed` / `accepted` |
| **3** | episode scope, negation and hypotheticals, ordinal cues, unit wording | **the hand-check sheets only** |

**Rung 3 is expected to lower `accepted`.** It exists to stop findings that should never have
been proposed, and no automatic counter can tell a dropped-wrong-extraction from a
missed-right-one. That distinction is what `supports_value` is, which is why the sheets are
filled against rung 3 and not against whichever rung scores highest in Step 8.

The note sample is picked by a fixed seed (`20260828`), so every rung sees the *same* notes.
Grades always come from the tables; extraction quality is what is being measured.

## Step 4 — Model choice and the Drive cache

Two independent things go wrong here, and the first run hit both.

**Quantization has to fit the GPU.** `jwang580/medgemma_27b_text_it` publishes one tag: F16,
54 GB. Colab hands out an A100-SXM4-**40 GB**. That model has never fit in VRAM — the earlier run
that "worked" was Ollama offloading it into system RAM and running on the CPU at **1.6 tok/s**
for 55 minutes. Below, quantization is chosen from detected VRAM so the model actually runs on
the GPU.

**Ollama mmaps its GGUF files, so it cannot serve them from Google Drive.** Drive FUSE does not
give mmap what it needs. Drive is therefore used as *cold storage only*: the model is copied to
local disk before the daemon ever touches it, and copied back after a pull.

> Whether that round-trip beats simply re-downloading is an open question — Drive FUSE reads are
> slow and Colab's link to HuggingFace is fast. Both are timed below. If the restore is slower
> than a fresh pull, set `USE_DRIVE = False` and stop paying for the cache.

```python
import json, os, pathlib, shutil, subprocess, time

USE_DRIVE = True          # Drive = cold storage only; serving is always from local disk

# --- quantization must fit VRAM -------------------------------------------
try:
    import torch
    VRAM_GB = torch.cuda.get_device_properties(0).total_memory / (1024 ** 3)
except Exception:
    VRAM_GB = 0.0

# The 16-bit path. Every published BF16/F16 GGUF of this model is split across two
# shards and Ollama cannot pull those, so Step 5 merges them and runs `ollama create`
# here in Colab. Nothing is downloaded from a registry. Set MODEL_OVERRIDE to None
# to fall back to the VRAM ladder below.
MODEL_OVERRIDE, MODEL_OVERRIDE_GB = "medgemma-27b-bf16", 54.0

# jwang580/medgemma_27b_text_it is the only single-file 16-bit tag Ollama can pull,
# and the 2026-08-30 run proved that conversion is BROKEN: 44 of 53 quotes came
# back as [UNK_BYTE_0xe29681...] byte-token wreckage, 83% of everything quoted.
# It is deliberately not reachable from this tier ladder. Use MODEL_OVERRIDE with a
# merged BF16 build if you need true 16-bit.
HF_REPO = "hf.co/unsloth/medgemma-27b-text-it-GGUF"
if MODEL_OVERRIDE:
    MODEL, MODEL_GB = MODEL_OVERRIDE, (MODEL_OVERRIDE_GB or 54.0)
elif VRAM_GB >= 34:
    MODEL, MODEL_GB = f"{HF_REPO}:Q8_0", 28.7      # near-lossless, and it actually works
else:
    MODEL, MODEL_GB = f"{HF_REPO}:Q4_K_M", 16.5
    if VRAM_GB < 22:
        print(f"!!  {VRAM_GB:.0f} GB VRAM - even Q4_K_M will partly offload to CPU and crawl.")
print(f"VRAM {VRAM_GB:.1f} GB  ->  {MODEL}  (~{MODEL_GB} GB)"
      + ("   [MODEL_OVERRIDE]" if MODEL_OVERRIDE else ""))

# --- speed: pin the model, and batch if the VRAM left over allows it -------
# Each parallel slot costs its own KV cache (~2 GB at 4096 ctx for this model).
# Weights + slots must leave the compute graph room, hence the 4 GB reserve.
# Pin the context length. Ollama's default has moved between versions, and KV is
# allocated per slot - letting it pick the model default (gemma3 declares 128K)
# would blow the headroom below. Prompts run ~1,000 tokens + 1,024 generated, so
# 4096 is ample and makes the arithmetic here true rather than hopeful.
NUM_CTX = 4096
KV_GB_PER_SLOT = 2.0                               # ~0.48 MB/token x 4096 for this model
headroom_gb = max(0.0, VRAM_GB - MODEL_GB - 4.0)   # 4 GB reserve for the compute graph
NUM_PARALLEL = max(1, min(4, int(headroom_gb // KV_GB_PER_SLOT)))
os.environ['OLLAMA_KEEP_ALIVE'] = '-1'             # never unload; a reload costs minutes
os.environ['OLLAMA_NUM_PARALLEL'] = str(NUM_PARALLEL)
os.environ['OLLAMA_CONTEXT_LENGTH'] = str(NUM_CTX)
os.environ['OLLAMA_FLASH_ATTENTION'] = '1'
print(f"headroom after weights {headroom_gb:.1f} GB  ->  OLLAMA_NUM_PARALLEL={NUM_PARALLEL} "
      f"x {NUM_CTX} ctx (~{NUM_PARALLEL * KV_GB_PER_SLOT:.0f} GB KV)")
if NUM_PARALLEL == 1:
    print("    no room to batch - the run stays sequential.")

# --- serving path: the BIGGEST local disk, never Drive ----------------------
# /root is not always the roomiest mount. Colab runtimes often carry a much larger
# scratch volume, and `ollama create` needs ~2.2x the model - the blob plus a
# llama-quantize compatibility rewrite - so picking the wrong mount fails late,
# after the merge, with "no space left on device".
_SKIP_FS = {"tmpfs", "devtmpfs", "squashfs", "proc", "sysfs", "cgroup", "cgroup2",
            "devpts", "fuse", "fuse.drive", "fuseblk"}
_disks = {}
for _line in open("/proc/mounts"):
    _p = _line.split()
    if len(_p) < 3:
        continue
    _target, _fs = _p[1], _p[2]
    if _fs in _SKIP_FS or "/drive" in _target:      # never serve off Drive: no mmap
        continue
    try:
        _u = shutil.disk_usage(_target)
    except OSError:
        continue
    if os.access(_target, os.W_OK) and _u.total > 20e9:
        _disks[_target] = _u.free

NEED_GB = MODEL_GB * 2.2
_ranked = sorted(_disks.items(), key=lambda kv: -kv[1])
print("local disks, most free first:")
for _t, _f in _ranked[:4]:
    print(f"   {_t:28s} {_f / 1e9:7.0f} GB free")

_root = _ranked[0][0] if _ranked else "/root"
SERVE_DIR = (pathlib.Path(_root) / "ollama_models" if _root != "/"
             else pathlib.Path("/root/.ollama/models"))
SERVE_DIR.mkdir(parents=True, exist_ok=True)
os.environ['OLLAMA_MODELS'] = str(SERVE_DIR)        # must be set before the daemon starts
local_free = shutil.disk_usage(SERVE_DIR).free / 1e9
print(f"serving from {SERVE_DIR}  ({local_free:.0f} GB free, need ~{NEED_GB:.0f} GB)")
if local_free < NEED_GB:
    print(f"!!  {local_free:.0f} GB free is under the ~{NEED_GB:.0f} GB `ollama create` "
          f"needs.")
    print( "    Set MODEL_OVERRIDE = None to fall back to Q8_0 (28.7 GB), or point")
    print( "    SERVE_DIR at a bigger mount from the list above and re-run this cell.")

# --- is a store REAL, or just a name? --------------------------------------
# A name in `ollama list`, and a manifest on disk, prove a model was REGISTERED.
# Neither says the 54 GB it points at is still there. A store whose blobs went
# missing - an interrupted copy, a wiped scratch disk, a cache saved without them -
# still lists the model; Ollama then cannot read the config blob, reports no
# capabilities, and refuses every generate with HTTP 400 "does not support
# generate". An undecodable GGUF gives that same 400 for an unrelated reason, so
# the status code cannot tell them apart. Checking the bytes here is what can.
def manifest_path(root, model):
    """Where Ollama keeps the manifest for `model` under a store root."""
    name, _, tag = model.partition(":")
    parts = name.split("/")
    if len(parts) == 1:                      # bare name -> the implied library path
        parts = ["registry.ollama.ai", "library"] + parts
    elif len(parts) == 2:                    # namespace/name -> the implied registry
        parts = ["registry.ollama.ai"] + parts
    return pathlib.Path(root, "manifests", *parts, tag or "latest")

def store_report(root, model):
    """-> (ok, [problems]). ok only when every blob the manifest names is present
    AND the size the manifest declares; a truncated blob is as unusable as an
    absent one, and a half-finished copy leaves exactly that."""
    man = manifest_path(root, model)
    if not man.is_file():
        return False, [f"no manifest at {man}"]
    try:
        doc = json.loads(man.read_text())
    except (OSError, ValueError) as e:
        return False, [f"manifest at {man} is unreadable: {e}"]
    named = [(l["digest"], l.get("size"))
             for l in [doc.get("config") or {}] + list(doc.get("layers") or [])
             if l.get("digest")]
    if not named:
        return False, [f"manifest at {man} names no blobs"]
    problems = []
    for digest, size in named:
        blob = pathlib.Path(root, "blobs", digest.replace(":", "-"))
        if not blob.exists():
            problems.append(f"missing blob {digest}")
        elif size and blob.stat().st_size != size:
            problems.append(f"short blob {digest} - {blob.stat().st_size / 1e9:.1f} GB "
                            f"of the {size / 1e9:.1f} GB the manifest declares")
    return not problems, problems

def store_intact(root, model):
    return store_report(root, model)[0]

# --- cold storage ----------------------------------------------------------
DRIVE_DIR = None
if USE_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive')                  # no-op if already mounted
    DRIVE_DIR = pathlib.Path('/content/drive/MyDrive/scogs_ollama_models')
    DRIVE_DIR.mkdir(parents=True, exist_ok=True)
    drive_gb = sum(f.stat().st_size for f in DRIVE_DIR.rglob('*') if f.is_file()) / 1e9

    # Two accepted layouts. A finished Ollama store is the fast path - Step 5 just
    # copies it. Raw shards mean Step 5 must merge and `ollama create` first.
    #
    # The store must hold THIS model. Drive very likely still has the broken
    # jwang580 build under manifests/, and a bare manifests/ check would happily
    # restore 54 GB of it before failing on a model that was never in there.
    _tag = MODEL.split(":")[0]
    STORE_MANIFEST = manifest_path(DRIVE_DIR, MODEL)
    HAVE_STORE, STORE_PROBLEMS = store_report(DRIVE_DIR, MODEL)
    # Three distinct states, and they need three different answers: a store for
    # some other model (dead weight), a store for THIS model with its blobs gone
    # (worse than nothing - restoring it produces a model that lists and cannot
    # run), and a store that is actually whole.
    HOLLOW_STORE = STORE_MANIFEST.exists() and not HAVE_STORE
    STALE_STORE = (DRIVE_DIR / "manifests").exists() and not STORE_MANIFEST.exists()
    _heads = sorted(DRIVE_DIR.glob("**/*-00001-of-*.gguf"))
    SHARD_HEAD = _heads[0] if _heads else None
    N_SHARDS = len(list(SHARD_HEAD.parent.glob(
        SHARD_HEAD.name.replace("00001-of-", "*-of-")))) if SHARD_HEAD else 0

    # An already-merged single-file GGUF beats re-merging. Anything with "-of-" in
    # the name is a shard; the largest of what is left is the merged model.
    _singles = [q for q in DRIVE_DIR.glob("**/*.gguf") if "-of-" not in q.name]
    MERGED_IN_DRIVE = max(_singles, key=lambda q: q.stat().st_size) if _singles else None

    print(f"Drive: {DRIVE_DIR}  ({drive_gb:.1f} GB held)")
    if STALE_STORE:
        stale = sorted(q.name for q in (DRIVE_DIR / "manifests").rglob("*")
                       if q.is_dir() and (q / "latest").exists())
        print(f"\n!!  Drive holds a store for {stale or 'another model'}, not {_tag}.")
        print( "    That is dead weight - it gets copied for nothing, or restored instead")
        print( "    of what you want. Clear it before running Step 5:")
        print( "        !rm -rf /content/drive/MyDrive/scogs_ollama_models/manifests")
        print( "        !rm -rf /content/drive/MyDrive/scogs_ollama_models/blobs\n")
    if HOLLOW_STORE:
        print(f"\n!!  Drive has a manifest for {_tag} but not the blobs behind it:")
        for _p in STORE_PROBLEMS[:4]:
            print(f"        {_p}")
        print( "    That is a name with nothing behind it, and restoring it gives a model")
        print( "    that lists fine and then fails every generate with 'does not support")
        print( "    generate'. Step 5 will ignore it and rebuild instead. Delete it so it")
        print( "    cannot be restored again:")
        print( "        !rm -rf /content/drive/MyDrive/scogs_ollama_models/manifests")
        print( "        !rm -rf /content/drive/MyDrive/scogs_ollama_models/blobs\n")
    if HAVE_STORE:
        print("  built Ollama store found, blobs verified -> Step 5 restores it directly.")
    elif MERGED_IN_DRIVE:
        print(f"  merged GGUF found: {MERGED_IN_DRIVE.name} "
              f"({MERGED_IN_DRIVE.stat().st_size / 1e9:.1f} GB)")
        print( "  -> Step 5 registers it with `ollama create`. No merge needed.")
        if SHARD_HEAD:
            print(f"     ({N_SHARDS} shards are also here and will be ignored - they are "
                  f"redundant now.)")
    elif SHARD_HEAD:
        print(f"  split GGUF found: {SHARD_HEAD.name}")
        print(f"  {N_SHARDS} shards -> Step 5 merges them, then runs `ollama create`.")
    else:
        print("\n!!  Drive holds no built store (manifests/), no merged .gguf, and no")
        print(f"    split GGUF (*-00001-of-*.gguf). Upload the BF16 shards into {DRIVE_DIR},")
        print( "    or set MODEL_OVERRIDE = None to use the VRAM ladder instead.")
```

## Step 5 — Get the model into the local store

**Nothing is downloaded from a registry.** The model comes from Drive, and this cell takes the
cheapest artifact that will work:

| Found | What happens | Cost |
|:--|:--|:--|
| local store already has it | nothing | free |
| `manifests/` + `blobs/` in Drive | copy to local disk | one copy |
| a merged single-file `.gguf` | `ollama create` reads it straight from Drive | one import |
| `*-00001-of-*.gguf` + siblings | merge, then `ollama create` | merge + import |

Ollama cannot load split GGUF files, which is why the merge exists. It runs against a pinned
CPU-only `llama.cpp` build — 16 MB, and a merge needs no GPU.

**The merge happens exactly once.** As soon as it finishes, the merged `.gguf` is copied to
Drive and the local copy is deleted — which both persists it and keeps peak local disk down.

That deletion is not tidiness. `ollama create` writes **two** more full-size files: the blob, then
a `llama-quantize` compatibility rewrite. Holding the merge locally at the same time needs about
**3× the model** — 162 GB here — which overruns the runtime disk. Creating from the Drive copy
keeps the peak at ~2×.

`SAVE_STORE_TO_DRIVE` additionally caches the built Ollama store, so later sessions skip
`ollama create` too. It is off by default because that is a third 54 GB copy in Drive.

A `.gguf` is only trusted if it is within 1% of the size the shards imply. A truncated merge or a
half-finished upload still has a valid header, so size is the only honest test — a short file is
reported and rebuilt rather than registered.

> ⚠️ `OLLAMA_MODELS`, `OLLAMA_KEEP_ALIVE` and `OLLAMA_NUM_PARALLEL` are read **once, when the
> daemon starts**. If a daemon is already up, this cell skips the start and your Step 4 changes
> do not take effect. Kill it first: `!pkill ollama`, then re-run Steps 4 and 5.

```python
import json, pathlib, shutil, subprocess, time, urllib.error, urllib.request

HOST = "http://localhost:11434"
LLAMA_BUILD = "b10701"            # pinned; ships llama-gguf-split
# The merge is persisted to Drive as a single .gguf either way, so the merge only
# ever happens once. Turning this on ALSO caches the built Ollama store, which
# skips `ollama create` on later runs too - at the cost of a third copy in Drive.
SAVE_STORE_TO_DRIVE = False
DELETE_SHARDS_AFTER_SAVE = False  # destructive; flip only once the merge is verified

def daemon_up(host=HOST, timeout=2):
    try:
        urllib.request.urlopen(host, timeout=timeout); return True
    except Exception:
        return False

def sh(*cmd, **kw):
    r = subprocess.run(cmd, capture_output=True, text=True, **kw)
    if r.returncode != 0:
        raise RuntimeError(f"{cmd[0]} failed ({r.returncode}):\n{(r.stderr or r.stdout)[-1500:]}")
    return r.stdout

def sh_stream(*cmd):
    """Run a command with its output visible as it goes.

    subprocess.run(capture_output=True) swallows every progress line until the command
    ends, so a running `ollama create` shows nothing for minutes and is indistinguishable
    from a hang. Read in chunks rather than lines: create writes progress with carriage
    returns, and line iteration would hold all of it back until the very end.
    """
    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                         text=True, bufsize=1)
    while True:
        chunk = p.stdout.read(256)
        if not chunk:
            break
        print(chunk, end="", flush=True)
    if p.wait() != 0:
        raise RuntimeError(f"{cmd[0]} failed ({p.returncode}) - see the output above")


def can_generate(tag, timeout=900):
    """-> (ok, why). A name in `ollama list` proves the model REGISTERED, not that it
    works. Ollama grants the `completion` capability by decoding the GGUF at request
    time, so a model that creates cleanly can still refuse every generate with
    'does not support generate'. One token settles it. The first call also loads the
    weights, hence the long timeout."""
    body = json.dumps({"model": tag, "prompt": "ok", "stream": False,
                       "options": {"num_predict": 1}}).encode("utf-8")
    req = urllib.request.Request(f"{HOST}/api/generate", data=body,
                                 headers={"Content-Type": "application/json"})
    try:
        urllib.request.urlopen(req, timeout=timeout)
        return True, ""
    except urllib.error.HTTPError as e:
        return False, f"HTTP {e.code} - {e.read().decode('utf-8', 'replace')[:300]}"
    except Exception as e:
        return False, str(e)


def log_tail(n=25):
    if OLLAMA_LOG.exists():
        lines = OLLAMA_LOG.read_text(errors="replace").splitlines()
        return "\n".join("      " + ln for ln in lines[-n:]) or "      (log is empty)"
    return f"      (no log at {OLLAMA_LOG} - the daemon was started outside this cell)"


def tree_bytes(d):
    return sum(q.stat().st_size for q in pathlib.Path(d).rglob('*') if q.is_file())

# 1. Ollama binary --------------------------------------------------------
# The installer extracts a .tar.zst and BAILS if zstd is missing - and it removes
# the previous install before it checks, so a missing zstd leaves you with no
# ollama at all. Install zstd first, then verify the binary really landed.
if shutil.which("ollama") is None:
    print("1/5 Installing zstd + Ollama ...")
    !sudo apt-get update -qq && sudo apt-get install -y -qq zstd
    !curl -fsSL https://ollama.com/install.sh | sh
    if shutil.which("ollama") is None:
        raise RuntimeError(
            "Ollama is still not on PATH after the install script ran. Read the output "
            "above - the usual cause is the zstd apt install failing. Do NOT work around "
            "this by hardcoding a path: if the binary is missing, a path to it is missing "
            "too, and every later step fails further from the cause.")
    print("    ollama installed.")
else:
    print("1/5 Ollama already installed - skipping.")

# 2. Get the weights onto local disk --------------------------------------
# Cheapest usable artifact wins: local store > Drive store > merged .gguf > shards.
merged = pathlib.Path('/content/merged.gguf')
create_from = None          # the .gguf `ollama create` should import, if any
merged_is_local = False     # True only when this cell wrote it, so only then delete it

# What a finished merge should weigh. Shards are the exact answer; without them
# fall back to the tier's declared size.
want = (sum(q.stat().st_size for q in SHARD_HEAD.parent.glob(
            SHARD_HEAD.name.replace("00001-of-", "*-of-")))
        if SHARD_HEAD else int(MODEL_GB * 1e9))

def gguf_complete(q):
    """A truncated merge still has a valid header, so size is the real test."""
    try:
        if q is None or not q.exists() or q.stat().st_size < want * 0.99:
            return False
        with open(q, "rb") as fh:
            return fh.read(4) == b"GGUF"
    except OSError:
        return False

# The local store gets the same blob check as Drive. Trusting the manifest alone
# is what turns a lost blob into a forty-minute detour: every later step says the
# model is present, and only /api/generate disagrees.
LOCAL_OK, LOCAL_PROBLEMS = store_report(SERVE_DIR, MODEL)
if not LOCAL_OK and manifest_path(SERVE_DIR, MODEL).exists():
    print(f"!!  The local store lists {MODEL}, but the bytes are not there:")
    for _p in LOCAL_PROBLEMS[:4]:
        print(f"        {_p}")
    print( "    Rebuilding it below - `ollama create` overwrites the manifest.\n")

if LOCAL_OK:
    print("2/5 Local store already has this model, blobs verified - skipping.")
elif HAVE_STORE:
    t0 = time.time()
    print(f"2/5 Restoring built store from Drive (~{drive_gb:.0f} GB) ...")
    sh("cp", "-r", f"{DRIVE_DIR}/.", str(SERVE_DIR))
    print(f"    restored in {time.time() - t0:.0f}s - no merge needed.")
elif gguf_complete(MERGED_IN_DRIVE):
    # Import straight from Drive. Copying it local first would cost a second 54 GB
    # for nothing - `ollama create` reads it once either way.
    print(f"2/5 Using the merged GGUF already in Drive "
          f"({MERGED_IN_DRIVE.stat().st_size / 1e9:.1f} GB) - no merge needed.")
    create_from = MERGED_IN_DRIVE
elif gguf_complete(merged):
    print(f"2/5 Reusing merged.gguf from earlier in this session "
          f"({merged.stat().st_size / 1e9:.1f} GB).")
    create_from, merged_is_local = merged, True
elif SHARD_HEAD:
    if MERGED_IN_DRIVE is not None:
        print(f"!!  {MERGED_IN_DRIVE.name} in Drive is only "
              f"{MERGED_IN_DRIVE.stat().st_size / 1e9:.1f} GB of ~{want / 1e9:.1f} GB - "
              f"an interrupted upload or merge. Ignoring it and rebuilding from shards; "
              f"delete it so it cannot be mistaken for the real thing.")
    # The merge itself only needs room for its own output; the expensive step is
    # `ollama create`, and that is checked separately just before it runs.
    free = shutil.disk_usage(SERVE_DIR).free / 1e9
    print(f"2/5 Merging {N_SHARDS} shards  (output ~{MODEL_GB:.0f} GB, {free:.0f} GB free)")
    if free < MODEL_GB * 1.1:
        raise RuntimeError(
            f"Only {free:.0f} GB free; the merge alone writes ~{MODEL_GB:.0f} GB. Use a "
            f"smaller quantization (MODEL_OVERRIDE = None falls back to Q8_0 at 28.7 GB), "
            f"or a runtime with more disk.")

    if not shutil.which("llama-gguf-split"):
        print(f"    fetching llama.cpp {LLAMA_BUILD} (CPU build, ~16 MB) ...")
        url = (f"https://github.com/ggml-org/llama.cpp/releases/download/{LLAMA_BUILD}"
               f"/llama-{LLAMA_BUILD}-bin-ubuntu-x64.tar.gz")
        !curl -sL -o /tmp/llama.tar.gz {url}
        !tar xzf /tmp/llama.tar.gz -C /tmp
        !cp /tmp/llama-{LLAMA_BUILD}/llama-gguf-split /usr/local/bin/
        !cp /tmp/llama-{LLAMA_BUILD}/*.so* /usr/local/lib/ 2>/dev/null
        !ldconfig
        !chmod +x /usr/local/bin/llama-gguf-split

    # llama-gguf-split refuses to overwrite, so a re-run after any later failure
    # would die on the short file the previous attempt left behind.
    if merged.exists():
        print(f"    discarding a partial merged.gguf "
              f"({merged.stat().st_size / 1e9:.1f} GB of ~{want / 1e9:.1f} GB).")
        merged.unlink()
    t0 = time.time()
    sh("llama-gguf-split", "--merge", str(SHARD_HEAD), str(merged))
    if not gguf_complete(merged):
        raise RuntimeError(
            f"The merge produced {merged.stat().st_size / 1e9:.1f} GB, short of the "
            f"~{want / 1e9:.1f} GB the shards hold. Do not register this - check the "
            f"shards are all present and local disk did not fill.")
    print(f"    merged {merged.stat().st_size / 1e9:.1f} GB in {time.time() - t0:.0f}s")
    create_from, merged_is_local = merged, True
else:
    raise RuntimeError("Drive holds no built store, no complete merged .gguf and no "
                       "split GGUF - see Step 4.")

# 2b. Hand a local merge to Drive, then drop it -----------------------------
# `ollama create` writes TWO more full-size files - the blob, then a llama-quantize
# compatibility rewrite - so holding the merge locally as well needs ~3x the model
# and overruns the runtime disk. Moving it to Drive persists it AND takes a full
# model off the local peak, which is the difference between fitting and not.
if merged_is_local and DRIVE_DIR:
    dst = DRIVE_DIR / f"{MODEL.split(':')[0]}.gguf"
    if gguf_complete(dst):
        print(f"    Drive already has {dst.name} - dropping the local copy.")
    else:
        t0 = time.time()
        print(f"    copying the merge to Drive as {dst.name} (one-time) ...")
        sh("cp", str(create_from), str(dst))
        if not gguf_complete(dst):
            dst.unlink(missing_ok=True)
            raise RuntimeError("The copy to Drive came out short and was removed. Check "
                               "your Drive quota, then re-run - the local merge is reused.")
        print(f"    copied in {time.time() - t0:.0f}s")
    merged.unlink()
    create_from, merged_is_local = dst, False
    print(f"    local merge removed; creating from Drive keeps peak disk at ~2x the model.")

# 3. Daemon (inherits the Step 4 environment) -----------------------------
# The log is the only place Ollama says WHY a model it accepted will not load
# ("couldn't decode ggml ..."). Sending it to DEVNULL made that unrecoverable and
# left the failure diagnosable only by guesswork, an hour downstream.
OLLAMA_LOG = pathlib.Path("/content/ollama.log")
if daemon_up():
    print("3/5 Daemon already running - skipping (Step 4 env NOT applied; !pkill ollama).")
    print(f"    NOTE: it was not started here, so {OLLAMA_LOG} may be stale or absent.")
else:
    print(f"3/5 Starting Ollama daemon (log -> {OLLAMA_LOG}) ...")
    _log = open(OLLAMA_LOG, "ab")
    ollama_proc = subprocess.Popen(["ollama", "serve"], stdout=_log, stderr=subprocess.STDOUT)
    for _ in range(45):
        time.sleep(2)
        if daemon_up(): break
    else:
        raise RuntimeError("Ollama daemon never came up on :11434")
    print("    daemon up.")

# 4. Register the merged file ---------------------------------------------
# Re-read the store: branch 2 may have restored it from Drive since LOCAL_OK was
# computed. `ollama list` would answer this question too, and answer it wrong - it
# reports names, and a name is what we already know we cannot trust.
if store_intact(SERVE_DIR, MODEL):
    print(f"4/5 {MODEL} already in the store, blobs verified - skipping create.")
elif create_from is not None:
    # ollama create writes the blob AND a llama-quantize compatibility rewrite,
    # so it needs about 2x the model free, on top of whatever the source costs.
    need = MODEL_GB * 2.2
    free = shutil.disk_usage(SERVE_DIR).free / 1e9
    print(f"4/5 ollama create {MODEL} from {create_from.name} "
          f"(needs ~{need:.0f} GB, {free:.0f} GB free) ...")
    if free < need:
        raise RuntimeError(
            f"Only {free:.0f} GB free where the store lives; `ollama create` writes the "
            f"blob and then a compatibility rewrite, about {need:.0f} GB in total. Free "
            f"space, or set MODEL_OVERRIDE = None to fall back to Q8_0 (28.7 GB).")
    t0 = time.time()
    mf = pathlib.Path('/content/Modelfile')
    mf.write_text(f"FROM {create_from}\n")
    sh_stream("ollama", "create", MODEL, "-f", str(mf))
    note = ""
    if merged_is_local:
        merged.unlink()                              # blobs/ holds it now
        note = "; local merged.gguf removed"
    print(f"\n    created in {time.time() - t0:.0f}s{note}")
else:
    _, _problems = store_report(SERVE_DIR, MODEL)
    raise RuntimeError(
        f"{MODEL} is not usable in the local store and there was nothing to build it "
        f"from:\n        " + "\n        ".join(_problems) + "\n"
        f"    If a Drive restore ran above, it came out incomplete - check Step 4's "
        f"Drive report, and the free space where the store lives.")

# 4b. Prove it. -------------------------------------------------------------
# Both branches above end with a NAME in `ollama list`, which is a registration
# check, not a working one. `ollama create` copies and hashes bytes; the tensor
# types are not exercised until something asks for a token. A BF16 GGUF that
# merges, creates and lists perfectly can still fail Ollama's capability check at
# request time - and then the first thing to touch it is the Step 6 probe, forty
# minutes and one model load later. One token here, where the cause is.
print("    verifying the model can generate (this loads the weights) ...")
ok, why = can_generate(MODEL)
if not ok:
    # Name the fault instead of guessing at it. HTTP 400 "does not support generate"
    # is Ollama saying "this model declares no capabilities", and it says that for
    # two unrelated reasons: it could not READ the config blob, or it could not
    # DECODE the GGUF. The store is the authority on the first and settles it
    # without ambiguity; only the log knows the second.
    store_ok, problems = store_report(SERVE_DIR, MODEL)
    log_text = OLLAMA_LOG.read_text(errors="replace") if OLLAMA_LOG.exists() else ""
    if not store_ok:
        why_it_failed = (
            "    The store is missing bytes its own manifest names:\n"
            + "\n".join(f"        {p}" for p in problems) + "\n\n"
            "    So this is NOT a BF16 problem - Ollama never reached a tensor. It could\n"
            "    not open the config blob, which leaves the model with no capabilities\n"
            "    and every generate refused. Fix the store, not the conversion:\n"
            "      1. Drop the empty registration and rebuild from the merged .gguf:\n"
            "           !pkill -f 'ollama serve'\n"
            f"           !rm -rf {manifest_path(SERVE_DIR, MODEL).parent}\n"
            "         then re-run Steps 4 and 5.\n"
            "      2. If Step 4 reported a store in Drive, that is where the hole came\n"
            "         from - delete its manifests/ and blobs/ too, or Step 5 restores\n"
            "         the same hole next time.\n"
            "      3. Check free space where the store lives. A create or a copy that\n"
            "         ran out of disk leaves exactly this.")
    elif "couldn't decode ggml" in log_text or "unsupported ggml type" in log_text:
        why_it_failed = (
            "    Every blob the manifest names is present and the right size, and the\n"
            "    log says the decoder rejected the GGUF. Ollama grants `completion` by\n"
            "    decoding the file when the request arrives, so it accepted these bytes\n"
            "    while copying them and refuses them while loading them. BF16 is the\n"
            "    usual reason - it is a late ggml type, and Ollama's Go decoder is a\n"
            "    different implementation from the llama.cpp one that merged the shards.\n"
            "      1. Upgrade Ollama and restart the daemon. The blob is already built\n"
            "         and full-size, so this needs no re-merge and no re-create:\n"
            "           !curl -fsSL https://ollama.com/install.sh | sh\n"
            "      2. Re-convert the shards to F16 rather than BF16.")
    else:
        why_it_failed = (
            "    The store is complete - every blob is present and the right size - and\n"
            "    the log does not mention a decode failure either. Read the tail below\n"
            "    before assuming a cause; this is not the known BF16 case.")
    raise RuntimeError(
        f"{MODEL} is registered but will not generate.\n"
        f"    Ollama said: {why}\n\n"
        f"{why_it_failed}\n\n"
        f"    Last lines of {OLLAMA_LOG}:\n"
        f"{log_tail()}")
print("    verified: it returns a token.")

# 5. Persist the built store so the merge never happens again -------------
if not (create_from is not None and SAVE_STORE_TO_DRIVE and DRIVE_DIR and not HAVE_STORE):
    if create_from is not None and not merged_is_local and DRIVE_DIR:
        print(f"5/5 The merge is already saved in Drive as {create_from.name}.")
        print( "    Later sessions reuse it and skip straight to `ollama create`.")
        if SHARD_HEAD:
            print(f"    The {N_SHARDS} shards (~{want / 1e9:.0f} GB) are redundant now - "
                  f"delete them,")
            print( "    or set DELETE_SHARDS_AFTER_SAVE = True to have this cell do it.")
        if DELETE_SHARDS_AFTER_SAVE and SHARD_HEAD:
            for q in SHARD_HEAD.parent.glob(SHARD_HEAD.name.replace("00001-of-", "*-of-")):
                q.unlink()
            print(f"    shards deleted, ~{want / 1e9:.0f} GB reclaimed (merge verified first).")
    else:
        print("5/5 Nothing to persist - skipping.")
else:
    local_bytes = tree_bytes(SERVE_DIR)
    try:
        drive_free = shutil.disk_usage(DRIVE_DIR).free
    except OSError:
        drive_free = None
    print(f"5/5 Saving the built store to Drive ({local_bytes / 1e9:.1f} GB) ...")
    if drive_free is not None and drive_free < local_bytes * 1.05:
        print(f"!!  Drive reports {drive_free / 1e9:.0f} GB free, need ~{local_bytes / 1e9:.0f} GB.")
        print( "    Skipping the save - the model still works this session, but the merge")
        print( "    will run again next time. Free space, or set DELETE_SHARDS_AFTER_SAVE")
        print( "    = True to trade the shards for the store.")
    else:
        t0 = time.time()
        sh("cp", "-r", f"{SERVE_DIR}/.", str(DRIVE_DIR))
        saved = tree_bytes(DRIVE_DIR / "blobs") if (DRIVE_DIR / "blobs").exists() else 0
        # Verify the copy the way Step 5 will later trust it. A save that reports
        # success on a manifest alone is what puts a hollow store in Drive, and the
        # next session restores it and fails on a model that cannot generate.
        drive_ok, drive_problems = store_report(DRIVE_DIR, MODEL)
        print(f"    saved in {time.time() - t0:.0f}s  "
              f"(blobs {saved / 1e9:.1f} GB, store {'verified' if drive_ok else 'INCOMPLETE'})")
        if not drive_ok:
            for _p in drive_problems[:4]:
                print(f"        {_p}")
            print("!!  The Drive copy is incomplete. Leave the shards in place and re-run")
            print("    this cell before trusting the cache - as saved it would restore a")
            print("    model that lists and will not run.")
        elif DELETE_SHARDS_AFTER_SAVE and SHARD_HEAD:
            for q in SHARD_HEAD.parent.glob(SHARD_HEAD.name.replace("00001-of-", "*-of-")):
                q.unlink()
            print(f"    shards deleted from Drive, ~{want / 1e9:.0f} GB reclaimed "
                  f"(store verified first).")
        else:
            print( "    Next session restores this store directly - no merge, no create.")
            print(f"    The shards and the merged .gguf in Drive are now redundant;")
            print( "    delete them once you have seen this store restore cleanly.")

print("\nlocal store:")
!ollama list
```

## Step 5b — build on the biggest local disk *(recovery, off by default)*

Run this **instead of Step 5** when Step 5 dies with `no space left on device`.

Two failures it fixes, both of which look identical to a hang:

| Cause | Symptom |
|:--|:--|
| `subprocess.run(capture_output=True)` swallows every progress line `ollama create` prints | the cell shows nothing until the command ends |
| `ollama create` reads and hashes the whole 54 GB **over Drive FUSE** before the daemon writes anything | 15–30 minutes of near-zero local disk activity |

So it stages the merged `.gguf` onto the fastest local disk as one bulk sequential
copy with a live MB/s and ETA, restarts the daemon with `OLLAMA_MODELS` pointed at
that disk, and streams `ollama create` through the `!` form. Headroom is 3.2× the
model — the staged copy alongside the blob and the compatibility rewrite — which is
why it wants the *biggest* mount, not `/root`.

Set `RUN_STEP_5B = True` to arm it.


```python
# Recovery path. Leave False unless Step 5 died with "no space left on device".
# Reuses MODEL / MODEL_GB / DRIVE_DIR from Steps 4-5 and rebinds SERVE_DIR.
RUN_STEP_5B = False

if not RUN_STEP_5B:
    print('Step 5b skipped (RUN_STEP_5B = False). Step 5 is the normal path.')
else:
    import os, pathlib, shutil, subprocess, time, urllib.request

    HOST      = "http://localhost:11434"
    MODEL_TAG = globals().get("MODEL", "medgemma-27b-bf16").split(":")[0]
    MODEL_GB  = float(globals().get("MODEL_GB", 54.0))
    NEED_GB   = MODEL_GB * 3.2          # staged copy + blob + compatibility rewrite

    # --- 1. every real local filesystem, most free first -----------------------
    SKIP = {"tmpfs", "devtmpfs", "squashfs", "proc", "sysfs", "cgroup", "cgroup2",
            "devpts", "fuse", "fuse.drive", "fuseblk"}
    disks = {}
    for line in open("/proc/mounts"):
        parts = line.split()
        if len(parts) < 3:
            continue
        target, fstype = parts[1], parts[2]
        if fstype in SKIP or "/drive" in target:
            continue
        try:
            u = shutil.disk_usage(target)
        except OSError:
            continue
        if os.access(target, os.W_OK) and u.total > 20e9:
            disks[target] = (u.free, u.total, fstype)

    print("writable local filesystems, most free first:")
    ranked = sorted(disks.items(), key=lambda kv: -kv[1][0])
    for target, (free, total, fstype) in ranked:
        print(f"   {target:30s} {fstype:12s} {free/1e9:7.0f} GB free / {total/1e9:.0f} GB")
    if not ranked:
        raise RuntimeError("Found no writable local filesystem. Run `!df -h` and set "
                           "SCRATCH_ROOT by hand below.")

    SCRATCH_ROOT = next((t for t, _ in ranked if "scratch" in t.lower()), ranked[0][0])
    # SCRATCH_ROOT = "/mnt/local-scratch"     # <- uncomment to force a specific disk
    free_gb = disks[SCRATCH_ROOT][0] / 1e9
    print(f"\nusing {SCRATCH_ROOT}  ({free_gb:.0f} GB free, need ~{NEED_GB:.0f} GB)")
    if free_gb < NEED_GB:
        raise RuntimeError(
            f"{SCRATCH_ROOT} has {free_gb:.0f} GB free but staging + `ollama create` needs "
            f"about {NEED_GB:.0f} GB. Pick another disk above, or set MODEL_OVERRIDE = None "
            f"in Step 4 to fall back to Q8_0 (28.7 GB).")

    STORE = pathlib.Path(SCRATCH_ROOT) / "ollama_models"
    STORE.mkdir(parents=True, exist_ok=True)

    # --- 2. find the merged .gguf ---------------------------------------------
    cands = []
    if globals().get("MERGED_IN_DRIVE"):
        cands.append(MERGED_IN_DRIVE)
    if globals().get("DRIVE_DIR"):
        cands += [q for q in pathlib.Path(DRIVE_DIR).glob("**/*.gguf") if "-of-" not in q.name]
    cands.append(pathlib.Path("/content/merged.gguf"))

    SRC = None
    for q in cands:
        try:
            if q.exists() and q.stat().st_size >= MODEL_GB * 1e9 * 0.99:
                with open(q, "rb") as fh:
                    if fh.read(4) == b"GGUF":
                        SRC = q
                        break
        except OSError:
            continue
    if SRC is None:
        raise RuntimeError("No complete merged .gguf found in Drive or at /content/merged.gguf.")
    print(f"source: {SRC}  ({SRC.stat().st_size / 1e9:.1f} GB)")

    # --- 3. stage it onto the fast disk, with progress -------------------------
    # `ollama create` reads and hashes the whole file before writing anything. Doing
    # that over Drive FUSE is 15-30 silent minutes. One bulk sequential copy first is
    # faster and, more to the point, visible.
    STAGED = pathlib.Path(SCRATCH_ROOT) / SRC.name
    total = SRC.stat().st_size
    if STAGED.exists() and STAGED.stat().st_size == total:
        print(f"already staged at {STAGED}")
    else:
        print(f"staging -> {STAGED}")
        t0, done, last = time.time(), 0, 0.0
        with open(SRC, "rb") as fi, open(STAGED, "wb") as fo:
            while chunk := fi.read(64 << 20):          # 64 MB at a time
                fo.write(chunk)
                done += len(chunk)
                if time.time() - last > 5:
                    el = time.time() - t0
                    rate = done / el / 1e6
                    eta = (total - done) / (done / el) if done else 0
                    print(f"   {done/1e9:5.1f} / {total/1e9:.1f} GB   "
                          f"{rate:5.0f} MB/s   eta {eta/60:4.1f} min", flush=True)
                    last = time.time()
        print(f"   staged {total/1e9:.1f} GB in {(time.time()-t0)/60:.1f} min")

    # --- 4. restart the daemon with its store on the big disk ------------------
    # OLLAMA_MODELS is read ONCE at startup, so the daemon has to be restarted.
    print("\nrestarting the daemon with the store on the big disk ...")
    subprocess.run(["pkill", "-f", "ollama serve"], capture_output=True)
    time.sleep(4)
    os.environ["OLLAMA_MODELS"] = str(STORE)
    os.environ.setdefault("OLLAMA_KEEP_ALIVE", "-1")
    os.environ.setdefault("OLLAMA_NUM_PARALLEL", str(globals().get("NUM_PARALLEL", 4)))
    os.environ.setdefault("OLLAMA_CONTEXT_LENGTH", str(globals().get("NUM_CTX", 4096)))
    os.environ.setdefault("OLLAMA_FLASH_ATTENTION", "1")
    SERVE_DIR = STORE            # keep later cells pointing at the right place

    subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    for _ in range(45):
        time.sleep(2)
        try:
            urllib.request.urlopen(HOST, timeout=2); break
        except Exception:
            pass
    else:
        raise RuntimeError("Ollama daemon never came up on :11434")
    print(f"    daemon up, OLLAMA_MODELS={STORE}")

    # --- 5. create, with the output actually visible ---------------------------
    mf = pathlib.Path("/content/Modelfile")
    mf.write_text(f"FROM {STAGED}\n")
    print(f"\nollama create {MODEL_TAG} - progress below, do not interrupt:")

    # The `!` form streams to the cell; subprocess.run(capture_output=True) hides it
    # all until the command ends, which is what made this look frozen.
    get_ipython().system(f'ollama create {MODEL_TAG} -f {mf}')

    # --- 6. verify -------------------------------------------------------------
    listed = subprocess.run(["ollama", "list"], capture_output=True, text=True).stdout
    print("\n" + listed)
    if MODEL_TAG not in listed:
        raise RuntimeError(f"{MODEL_TAG} is not in the store - read the create output above.")
    print(f"store: {STORE}  ({shutil.disk_usage(STORE).free / 1e9:.0f} GB still free)")
    print(f"You can reclaim {total/1e9:.0f} GB now:   !rm {STAGED}")
    print("This disk is wiped when the runtime ends; Steps 6 and 7 work for this session.")
```

## Step 5c — cache the built store in Drive *(recovery, off by default)*

Run after Step 5b succeeds. Copies the built Ollama store to Drive so later sessions
skip **both** the merge and the create.

It prunes orphaned blobs first — the ones no manifest references, left behind by
failed attempts — by reading every manifest and keeping only the digests they name,
so nothing still in use is ever deleted. Then it copies with progress, verifies file
sizes and the manifest before trusting the copy, and only then offers to delete the
now-redundant shards and merged `.gguf`.

Set `RUN_STEP_5C = True` to arm it.


```python
# Recovery path. Leave False unless you ran Step 5b and want the built store
# cached in Drive so future sessions skip both the merge and the create.
RUN_STEP_5C = False

if not RUN_STEP_5C:
    print('Step 5c skipped (RUN_STEP_5C = False). Step 5 persists the store on its own.')
else:
    import json, os, pathlib, shutil, time

    PRUNE_ORPHANS  = True     # delete blobs no manifest references (failed-attempt leftovers)
    DELETE_SHARDS  = False    # delete the split shards from Drive after the store verifies
    DELETE_MERGED  = False    # delete the merged .gguf from Drive after the store verifies

    STORE = pathlib.Path(globals().get("SERVE_DIR", "/mnt/local-scratch/ollama_models"))
    DRIVE = pathlib.Path(globals().get("DRIVE_DIR", "/content/drive/MyDrive/scogs_ollama_models"))
    TAG   = globals().get("MODEL", "medgemma-27b-bf16").split(":")[0]

    if not (STORE / "manifests").exists():
        raise RuntimeError(f"No Ollama store at {STORE}. Run Step 5b first.")

    # --- 1. what is actually in the store? ------------------------------------
    def referenced_digests(store):
        """Every blob named by any manifest, so we never delete one still in use."""
        keep = set()
        for mf in (store / "manifests").rglob("*"):
            if not mf.is_file():
                continue
            try:
                data = json.loads(mf.read_text())
            except Exception:
                continue
            for layer in (data.get("layers") or []) + [data.get("config") or {}]:
                d = layer.get("digest", "")
                if d.startswith("sha256:"):
                    keep.add("sha256-" + d.split(":", 1)[1])
        return keep

    blobs = {q.name: q.stat().st_size for q in (STORE / "blobs").glob("*") if q.is_file()}
    keep  = referenced_digests(STORE)
    orphans = {n: s for n, s in blobs.items() if n not in keep}

    print(f"store {STORE}")
    print(f"  blobs        {len(blobs):3d}  {sum(blobs.values())/1e9:7.1f} GB")
    print(f"  referenced   {len(keep & set(blobs)):3d}  "
          f"{sum(s for n, s in blobs.items() if n in keep)/1e9:7.1f} GB")
    print(f"  orphaned     {len(orphans):3d}  {sum(orphans.values())/1e9:7.1f} GB"
          + ("   <- leftovers from the failed attempts" if orphans else ""))

    if orphans and PRUNE_ORPHANS:
        for n in orphans:
            (STORE / "blobs" / n).unlink()
        print(f"  pruned {len(orphans)} orphan blob(s), "
              f"{sum(orphans.values())/1e9:.1f} GB reclaimed")

    need = sum(q.stat().st_size for q in STORE.rglob("*") if q.is_file())
    print(f"\nto copy: {need/1e9:.1f} GB")

    # --- 2. room in Drive? -----------------------------------------------------
    drive_used = sum(q.stat().st_size for q in DRIVE.rglob("*") if q.is_file())
    try:
        drive_free = shutil.disk_usage(DRIVE).free
    except OSError:
        drive_free = None
    print(f"Drive holds {drive_used/1e9:.1f} GB"
          + (f", {drive_free/1e9:.0f} GB free" if drive_free is not None else ""))

    shards = sorted(DRIVE.glob("**/*-of-*.gguf"))
    mergedf = [q for q in DRIVE.glob("**/*.gguf") if "-of-" not in q.name]
    if drive_free is not None and drive_free < need * 1.05:
        msg = [f"Drive has {drive_free/1e9:.0f} GB free but the store needs {need/1e9:.0f} GB."]
        if shards:
            msg.append(f"The {len(shards)} shards (~{sum(q.stat().st_size for q in shards)/1e9:.0f} GB) "
                       f"are redundant - the merged .gguf is built and verified. Set "
                       f"DELETE_SHARDS = True, or delete them by hand, then re-run.")
        raise RuntimeError(" ".join(msg))

    # --- 3. copy, with progress ------------------------------------------------
    def copy_with_progress(src, dst):
        dst.parent.mkdir(parents=True, exist_ok=True)
        total, done, t0, last = src.stat().st_size, 0, time.time(), 0.0
        with open(src, "rb") as fi, open(dst, "wb") as fo:
            while chunk := fi.read(64 << 20):
                fo.write(chunk); done += len(chunk)
                if total > 2e9 and time.time() - last > 5:
                    el = time.time() - t0
                    print(f"      {done/1e9:5.1f} / {total/1e9:.1f} GB   "
                          f"{done/el/1e6:5.0f} MB/s   eta {(total-done)/(done/el)/60:4.1f} min",
                          flush=True)
                    last = time.time()

    print(f"\ncopying {STORE} -> {DRIVE}")
    t0 = time.time()
    for src in sorted(STORE.rglob("*")):
        if not src.is_file():
            continue
        dst = DRIVE / src.relative_to(STORE)
        if dst.exists() and dst.stat().st_size == src.stat().st_size:
            continue
        print(f"   {src.relative_to(STORE)}  ({src.stat().st_size/1e9:.1f} GB)")
        copy_with_progress(src, dst)
    print(f"copied in {(time.time()-t0)/60:.1f} min")

    # --- 4. verify before trusting it ------------------------------------------
    ok = True
    for src in STORE.rglob("*"):
        if not src.is_file():
            continue
        dst = DRIVE / src.relative_to(STORE)
        if not dst.exists() or dst.stat().st_size != src.stat().st_size:
            print(f"!!  MISMATCH {src.relative_to(STORE)}")
            ok = False
    manifest_ok = (DRIVE / "manifests" / "registry.ollama.ai" / "library" / TAG).exists()
    print(f"\nverify: files {'ok' if ok else 'FAILED'}, manifest for {TAG} "
          f"{'ok' if manifest_ok else 'MISSING'}")
    if not (ok and manifest_ok):
        raise RuntimeError("The Drive copy is incomplete. Do NOT delete anything; re-run.")

    print(f"\nDone. Next session, Step 5 finds this store and just copies it back - "
          f"no merge, no create.")

    # --- 5. optional cleanup, only now that the store is verified --------------
    for flag, files, label in ((DELETE_SHARDS, shards, "shards"),
                               (DELETE_MERGED, mergedf, "merged .gguf")):
        if flag and files:
            freed = sum(q.stat().st_size for q in files)
            for q in files:
                q.unlink()
            print(f"deleted {label} from Drive, {freed/1e9:.0f} GB reclaimed")
        elif files:
            print(f"{label} still in Drive (~{sum(q.stat().st_size for q in files)/1e9:.0f} GB) "
                  f"- redundant now; delete by hand or set the flag above.")
```

## Step 6 — Preflight gate ⛔

**This cell can halt the notebook, and that is the point.** Two failures are invisible until they
have already ruined a run:

| Failure | Symptom if unchecked |
|:--|:--|
| A corrupt GGUF conversion emitting `[UNK_BYTE_…]` byte tokens | every quote fails verbatim match; reads as a hallucinating model |
| The model sitting on CPU instead of the A100 | ~1.6 tok/s, an hour-long run, meaningless cost numbers |

It also records the **real quantization and digest** — §2 provenance requires the digest, and the
previous run stored `model_digest: 1`, which is a file-type enum, not a hash.

Loosening the quote matcher to tolerate corrupt tokens is *not* the fix. That matcher is the
project's entire hallucination defence; the fix is to serve weights that are not broken.

```python
import json, re, subprocess, urllib.error, urllib.request

ARTIFACTS = re.compile(r"\[UNK_BYTE_|▁")
fatal = []

# ---------------------------------------------------------------- provenance
tags = json.loads(urllib.request.urlopen(f"{HOST}/api/tags").read())["models"]
entry = next((m for m in tags if m["name"].startswith(MODEL.split(":")[0])), None)
if entry is None:
    raise RuntimeError(f"{MODEL} is not in the local cache - re-run Step 5.")

det = entry.get("details", {})
QUANT = det.get("quantization_level", "unknown")
DIGEST = entry.get("digest", "")
SIZE_GB = entry.get("size", 0) / 1e9
print("PROVENANCE")
print(f"  tag            {entry['name']}")
print(f"  family/params  {det.get('family')} / {det.get('parameter_size')}")
print(f"  format/quant   {det.get('format')} / {QUANT}")
print(f"  size on disk   {SIZE_GB:.1f} GB")
print(f"  digest         {DIGEST}")
if QUANT.upper() not in {"F16", "BF16", "F32"}:
    print(f"\n  NOTE: this is {QUANT}, NOT full 16-bit. Report it as {QUANT} in any results table.")

# ------------------------------------------------- does it even fit? (cheap)
# Check this BEFORE asking the model to generate. Ollama accepts a pull it cannot
# load and then fails /api/generate with a bare HTTP 500, which says nothing.
print(f"\nCAPACITY   model {SIZE_GB:.1f} GB vs VRAM {VRAM_GB:.1f} GB")
if VRAM_GB and SIZE_GB > 0.9 * VRAM_GB:
    fatal.append(
        f"{SIZE_GB:.1f} GB of weights will not fit {VRAM_GB:.1f} GB of VRAM. Ollama will "
        f"fail the load with an opaque 500. Pull a smaller quantization instead - "
        f"hf.co/unsloth/medgemma-27b-text-it-GGUF has Q8_0 (28.7 GB) and Q4_K_M (16.5 GB) - "
        f"and `ollama rm {entry['name']}` to reclaim the cache space.")
else:
    print("  fits.")

# ------------------------------------------------- warm up, then GPU split
# TEN probes, not one. The 2026-08-30 run passed this gate on a single sentence and
# then produced byte-token wreckage in 28 of 80 replies - corruption is intermittent,
# so n=1 proves nothing. Any artifact in any probe is fatal.
PROBES = [
    "The patient was started on norepinephrine and FiO2 was escalated to 60%.",
    "Creatinine at the time of explant was 0.9 mg/dl.",
    "He was treated with hydroxyurea 500 mg daily for two years.",
    "Temperature on admission was 38.9 C with a heart rate of 122 bpm.",
    "A 22-year-old male presented with acute chest syndrome.",
    "Hemoglobin dropped from 9.4 g/dL to 6.1 g/dL over 48 hours.",
    "Exchange transfusion was performed; no further vaso-occlusive episodes occurred.",
    "The patient required non-invasive ventilation (BiPAP) for 36 hours.",
    "Pain was resolved by analgesics, but he never sought medical treatment.",
    "Renal replacement therapy was initiated on hospital day 4.",
]

def probe_once(text):
    req = urllib.request.Request(
        f"{HOST}/api/generate",
        data=json.dumps({
            "model": MODEL, "stream": False, "format": "json",
            "prompt": ('Copy the sentence below exactly, character for character, as JSON '
                       '{"quote": "<sentence>"}. Reply with JSON only.\n\n' + text),
            "options": {"temperature": 0, "seed": 0, "num_predict": 128},
        }).encode(),
        headers={"Content-Type": "application/json"})
    try:
        return json.loads(urllib.request.urlopen(req, timeout=900).read())
    except urllib.error.HTTPError as e:
        # The body is where Ollama puts the real reason. Never swallow it.
        body = e.read().decode("utf-8", "replace")[:600]
        raise RuntimeError(f"Ollama returned HTTP {e.code} on /api/generate.\n"
                           f"Ollama said: {body}\n"
                           f"Run `!ollama logs` or check `!ollama ps` for more.") from None

if not fatal:
    print(f"\nTOKENIZER INTEGRITY   ({len(PROBES)} probes)")
    dirty, tok_rates, first_bad = [], [], None
    for n, text in enumerate(PROBES, 1):
        resp = probe_once(text)
        raw = resp.get("response", "")
        dur = resp.get("eval_duration", 0) or 1
        tok_rates.append(resp.get("eval_count", 0) / (dur / 1e9))
        if ARTIFACTS.search(raw):
            dirty.append(n)
            if first_bad is None:
                first_bad = raw
        print(f"  probe {n:2d}/{len(PROBES)}  "
              + ("CORRUPT" if n in dirty else "clean  ")
              + f"  {tok_rates[-1]:5.1f} tok/s")

    if dirty:
        print(f"\n  first corrupt reply: {first_bad[:200]}")
        fatal.append(
            f"{len(dirty)} of {len(PROBES)} probes came back with SentencePiece byte-token "
            f"wreckage ([UNK_BYTE_...] / U+2581). The served GGUF conversion is broken. "
            f"Note the harness's normalize() STRIPS these before matching, so a run would "
            f"still report a high quote-verified rate - measured on repaired text, not on "
            f"what the model said. Switch to a clean conversion; do NOT loosen the matcher.")
    else:
        print(f"  all {len(PROBES)} clean - no [UNK_BYTE_...] or U+2581 artifacts.")

    print("\nPLACEMENT (ollama ps)")
    subprocess.run(["ollama", "ps"])
    tok_s = sum(tok_rates) / len(tok_rates)
    print(f"  mean probe throughput  {tok_s:.1f} completion tok/s "
          f"(min {min(tok_rates):.1f}, max {max(tok_rates):.1f})")
    if tok_s < 5:
        fatal.append(f"Throughput {tok_s:.1f} tok/s means the model is largely on CPU. "
                     f"The PROCESSOR column above must read 100% GPU.")

# ------------------------------------------------------------------- verdict
print("\n" + "=" * 72)
if fatal:
    for f in fatal:
        print("FATAL: " + f)
    print("=" * 72)
    raise RuntimeError("Preflight failed - fix the above before spending an hour on a run.")
print("Preflight passed. Safe to run the harness.")
print("=" * 72)
```

## Step 7 — Run the ladder

Every rung runs the **same notes** (seed `20260828`), the same four outcomes and the same
`--repeat`, changing only `--prompt-stage`. That is the whole point: a difference between two
result files is then attributable to the prompt and to nothing else.

Set `STAGES_TO_RUN = ["0"]` on a first pass if you only want the baseline. `SKIP_DONE` leaves
finished rungs alone, so a dropped Colab session resumes instead of starting over.

**Cost.** The prompt grows with the rung — roughly 580 tokens per pair at rung 0 and 1,350 at
rung 3, before the note. Prefill on an A100 absorbs most of that; budget about 15 minutes per
rung at `--concurrency 1`, less if you set `MEASURE_CONSISTENCY = False`.

`--repeat 2` at `--concurrency 1` keeps `run_to_run_consistency_pct` meaningful. It measures
**determinism**, not agreement — at temperature 0 with a fixed seed the repeats should be
identical, and a number below 98% means something in the serving path is non-deterministic,
not that the model is unsure. It is not an ensemble and does not vote.

```python
import os, pathlib, subprocess, time

# Trim this to re-run a single rung. Order matters only for your patience.
STAGES_TO_RUN = ["0", "1", "2a", "2b", "3"]
SKIP_DONE     = True          # leave a rung alone if its result file already exists
NOTES         = 20
REPEAT        = 2

# True  -> concurrency 1: run_to_run_consistency_pct stays meaningful (P11 gate).
# False -> batch it: several times faster, consistency confounded (recorded in provenance).
MEASURE_CONSISTENCY = True
CONCURRENCY = 1 if MEASURE_CONSISTENCY else NUM_PARALLEL

LADDER_DIR = "results/ladder"
pathlib.Path(LADDER_DIR).mkdir(parents=True, exist_ok=True)
def stage_path(s): return f"{LADDER_DIR}/stage_{s}.json"

print(f"ladder: {' -> '.join(STAGES_TO_RUN)}   notes={NOTES}  repeat={REPEAT}  "
      f"concurrency={CONCURRENCY}"
      + ("" if CONCURRENCY == 1 else "   (consistency will be flagged as confounded)"))

for s in STAGES_TO_RUN:
    out = stage_path(s)
    if SKIP_DONE and os.path.exists(out):
        print(f"\n=== stage {s}: already at {out}, skipping (SKIP_DONE) ===")
        continue
    print(f"\n{'=' * 74}\n=== stage {s} -> {out}\n{'=' * 74}", flush=True)
    t0 = time.time()
    cmd = (f"python scripts/experiments/medgemma_extraction.py"
           f" --tier full --model {MODEL} --backend ollama"
           f" --cohort scd_primary --stratify --holdout-frac 0.25"
           f" --notes {NOTES} --repeat {REPEAT} --concurrency {CONCURRENCY}"
           f" --prompt-stage {s} --out {out}")
    !{cmd}
    if not os.path.exists(out):
        raise SystemExit(f"stage {s} produced no result file - stop here rather than "
                         f"comparing a ladder with a hole in it")
    print(f"--- stage {s} finished in {(time.time() - t0) / 60:.1f} min")

print("\nDone. Step 8 compares whatever rungs are on disk.")
```

## Step 8 — The ladder ⭐

One row per rung. Read it in three passes:

1. **Rung 1 is a correctness check, not a result.** `null_placeholder` and `unknown_feature`
   should be **0** — the grammar makes them unreachable. If they are not, the schema is not
   reaching the decoder and everything above this row is measuring the wrong thing.
2. **Rungs 2a and 2b are results.** Watch `graded` rising against `grade_set`: 16 pairs were
   stuck one step short for want of `death_attributed`, and that is what 2a is for.
3. **Rung 3 is not judged here.** `accepted` going down is the intended behaviour. The column
   that decides it is `supports_value`, in Step 12.

`quote_verified` is a **grounding** number throughout — the quoted words are in the note. It
never asks whether they support the value. Nothing on this table is precision.

```python
import json, os, glob
import pandas as pd

LADDER_DIR = "results/ladder"
found = sorted(glob.glob(f"{LADDER_DIR}/stage_*.json"),
               key=lambda p: ["0", "1", "2a", "2b", "3"].index(
                   os.path.basename(p)[len("stage_"):-len(".json")]))
assert found, "No ladder result files - run Step 7 first."

rows, uid_sets = [], {}
for path in found:
    with open(path, encoding="utf-8") as f:
        d = json.load(f)
    prov, met = d["provenance"], d["automated_metrics"]
    st = prov.get("prompt_stage", "?")
    uid_sets[st] = tuple(sorted(r["patient_uid"] for r in d["detailed_records"]))
    # Grade counts live at the top level, not inside automated_metrics.
    g = d.get("grade_status") or {}
    rows.append({
        "stage": st,
        "proposed": met["proposed"],
        "accepted": met["accepted"],
        "null%": met["null_placeholder_pct"],
        "unfound%": met["hallucinated_pct_of_quoted"],
        "unknown_feat": met.get("unknown_feature", 0),
        "bad_value": met.get("invalid_value_count", 0),
        "pres_quoted%": met.get("present_quoted_pct", float("nan")),
        "retries": met.get("content_retries", 0),
        "graded": g.get("graded", 0),
        "grade_set": g.get("grade_set", 0),
        "refuted": g.get("refuted", 0),
        "absent": g.get("absent", 0),
        "cannot": g.get("cannot_grade", 0),
        "consistency%": met.get("run_to_run_consistency_pct"),
        "min": round(d["profiling"]["total_wall_clock_sec"] / 60, 1),
    })

lad = pd.DataFrame(rows).set_index("stage")

# The comparison is only worth reading if every rung saw the same notes. Selection is
# seeded, so this should never fire - but a changed --notes or --cohort between rungs
# would silently turn the ladder into five unrelated runs.
distinct = set(uid_sets.values())
if len(distinct) > 1:
    print("!! STOP: the rungs did NOT see the same notes. Sample differs between:")
    for st, uids in uid_sets.items():
        print(f"     stage {st}: {len(uids)} notes, first {uids[:3]}")
    print("   Re-run the ladder with identical --notes / --cohort before comparing.\n")
else:
    print(f"All {len(lad)} rungs ran the same {len(next(iter(distinct)))} notes.\n")

pd.set_option("display.width", 200)
display(lad)

if "0" in lad.index and len(lad) > 1:
    delta = lad.drop(columns=["consistency%"]).astype(float).subtract(
        lad.drop(columns=["consistency%"]).astype(float).loc["0"])
    print("\nChange from stage 0:")
    display(delta.drop(index="0").round(1))

# ---- the one automatic check that is pass/fail rather than directional ----
if "1" in lad.index:
    r = lad.loc["1"]
    bad = {k: r[k] for k in ("null%", "unknown_feat", "bad_value") if r[k] > 0}
    if bad:
        print(f"\n!! Stage 1 should make these unreachable, and did not: {bad}")
        print("   The JSON schema is not reaching the decoder. Check that Ollama is recent")
        print("   enough to accept a schema object in `format` (a bare \"json\" string only")
        print("   guarantees the reply parses). Rungs above this one are not interpretable")
        print("   until it is fixed.")
    else:
        print("\nStage 1 OK: null placeholders, unknown features and bad values all 0 -")
        print("the grammar made them unreachable rather than the tally counting them.")
```

## Step 10 — Results for that rung

Grounding is reported against **both denominators, always together**. A finding with no quote is a
*prompt-compliance* failure, not a grounding result; quoting one number without the other is how a
run reads as either 7% or 96% depending on which you pick.

```python
import json, os
import pandas as pd

with open(RESULT_PATH, encoding="utf-8") as f:
    data = json.load(f)

prov, prof = data["provenance"], data["profiling"]
met, runs = data["automated_metrics"], data["runs"]

print("=" * 74)
print(f"MedGemma extraction report   stage={prov.get('prompt_stage', '?')}  "
      f"tier={prov['tier']}  cohort={prov.get('cohort', 'loose')}")
print(f"weights={prov['weights']}   served_as={prov['served_as']}   backend={prov['backend']}")
print(f"digest={prov.get('model_digest')}   quant={prov.get('quant')}   {prov['timestamp']}")
print(f"run_id={prov.get('run_id', '(none - re-run with the current harness)')}")
print("=" * 74)

# ---- cohort composition: absences the sample forced, not the model ----
recs = data["detailed_records"]
n_primary = sum(1 for r in recs if r.get("scd_primary"))
print(f"\nCOHORT   {prov.get('cohort', 'loose')}   "
      f"{n_primary}/{len(recs)} selected notes are SCD-primary")
if n_primary < len(recs):
    print(f"  {len(recs) - n_primary} note(s) only MENTION sickle cell - trait, a denial, a relative's")
    print( "  diagnosis, or cardiology 'SCD' = sudden cardiac death. Each contributes")
    print(f"  {len(prov['outcomes'])} pairs that can only score `absent`, diluting every rate below.")
    print( "  Re-run Step 7 with --cohort scd_primary to remove them.")
else:
    print("  Every selected note is about the disease - no absent-by-construction dilution.")

prop, quoted = met["proposed"], met["quoted"]
print(f"\nGROUNDING   ({prop} proposed findings)")
print(f"  null placeholders   {met['null_placeholder']:4d}   {met['null_placeholder_pct']:5.1f}% of proposed"
      + ("   <- omission rule ignored" if met['null_placeholder_pct'] > 5 else ""))
print(f"  quote verified      {met['quote_verified']:4d}   "
      f"{met['quote_verified_pct_of_quoted']:5.1f}% of quoted | {met['quote_verified_pct']:.1f}% of all")
print(f"  quote not in note   {met['quote_unfound']:4d}   "
      f"{met['hallucinated_pct_of_quoted']:5.1f}% of quoted | {met['hallucinated_quote_pct']:.1f}% of all"
      "   <- the hallucination rate")
print("\n  Quote-verified is a GROUNDING check: it asks whether the quoted words are in")
print("  the note, never whether they support the value. 'needed increasing oxygen by")
print("  nasal cannula' verifies perfectly behind fio2_pct=21. Precision is Step 10.")
if met.get("unit_converted") or met.get("unit_mismatch") or met.get("quote_value_mismatch"):
    print(f"\nUNIT GUARD   (a number is read from its own quote, in the unit the note wrote it in)")
    print(f"  converted into the schema's unit   {met.get('unit_converted', 0):4d}")
    print(f"  unconvertible unit, rejected       {met.get('unit_mismatch', 0):4d}")
    print(f"  value not the quote's number       {met.get('quote_value_mismatch', 0):4d}")
if met.get("value_conflicts"):
    print(f"\nVALUE CONFLICTS   {met['value_conflicts']} feature(s) had several verified values and no")
    print( "  aggregation rule in the schema. Withheld from grading rather than guessed at;")
    print( "  they are listed in Step 10 and are the highest-yield rows a reviewer can read.")
if met.get("tokenizer_artifacts"):
    print(f"\n  !! {met['tokenizer_artifacts']} quotes carry corrupt GGUF byte tokens - "
          "these numbers are not a clean measurement.")

def band(v, good, workable):
    return "GOOD" if v >= good else ("WORKABLE" if v >= workable else "CONCERNING")

# A consistency number measured under batching cannot be read as a model property.
# ...and at concurrency 1 with greedy decoding, 100% is the expected result. It is a
# smoke test for a nondeterministic serving stack, not evidence about the model.
confounded = prov.get("consistency_confounded_by_batching", False)
cons = met.get("run_to_run_consistency_pct")
if cons is None:
    consistency_verdict, cons_txt = "NOT MEASURED", "-"
else:
    cons_txt = f"{cons:.1f}%"
    if confounded:                       consistency_verdict = "CONFOUNDED"
    elif cons < 98:                      consistency_verdict = band(cons, 98, 90)
    elif prov.get("concurrency", 1) == 1: consistency_verdict = "EXPECTED (near-tautology)"
    else:                                consistency_verdict = "GOOD"

gate = pd.DataFrame([
    ("Quote-verified % (of quoted)", f"{met['quote_verified_pct_of_quoted']:.1f}%",
     "≥95 / 85-95 / <85", band(met["quote_verified_pct_of_quoted"], 95, 85)),
    ("Run-to-run consistency", cons_txt, "≥98 / 90-98 / <90", consistency_verdict),
    ("Invalid-value rate", f"{met['invalid_value_pct']:.1f}%",
     "≤2 / 2-10 / >10", band(-met["invalid_value_pct"], -2, -10)),
    ("Unparseable replies", f"{met['unparseable_replies']}",
     "0", "GOOD" if met["unparseable_replies"] == 0 else "CONCERNING"),
    ("Null-placeholder rate", f"{met['null_placeholder_pct']:.1f}%",
     "≤5", "GOOD" if met["null_placeholder_pct"] <= 5 else "CONCERNING"),
], columns=["P11 automated metric", "Value", "Good / Workable / Concerning", "Verdict"])
display(gate)
if confounded:
    print(f"!! Consistency was measured at concurrency {prov.get('concurrency')}. Batching can flip a\n"
          f"   token at temperature 0 on its own, so that number is not a model property and the\n"
          f"   P11 gate cannot use it. Re-run with MEASURE_CONSISTENCY = True to get a clean one.")

cost = pd.DataFrame([
    ("Notes × outcomes", f"{prov['notes_count']} × {len(prov['outcomes'])}"),
    ("Concurrency", f"{prov.get('concurrency', 1)}"),
    ("Wall clock", f"{prof['total_wall_clock_sec']:.0f} s"),
    ("Per note (these outcomes)", f"{prof['sec_per_note']:.1f} s"),
    ("Throughput", f"{prof['completion_tokens_per_sec']:.1f} completion tok/s"),
    ("Prompt tokens / note", f"{prof['prompt_tokens_per_note']:,}"),
    ("Extrapolated to all 53 outcomes",
     f"{prof['sec_per_note'] * 53 / len(prov['outcomes']) / 60:.0f} min/note, "
     f"{prof['prompt_tokens_per_note'] * 53 // len(prov['outcomes']):,} prompt tok/note"),
], columns=["Cost", "Value"])
display(cost)

status = data["grade_status"]
total = sum(status.values())
display(pd.DataFrame(
    [(k, v, f"{100 * v / total:.1f}%") for k, v in sorted(status.items(), key=lambda kv: -kv[1])],
    columns=["SCOGS status", "Note-outcome pairs", "Share"]))

# `refuted` is split out of `absent`: the model DID evidence the outcome and the
# decision tables overruled the call. Pooling the two sends a reviewer to confirm an
# absence that the rule engine, not the model, produced.
COLS = ["graded", "grade_set", "cannot_grade", "absent", "refuted", "not_applicable"]
n_notes = prov["notes_count"]

print(f"\nPER OUTCOME (n={n_notes} each) - a pooled number hides this shape entirely.")
print("An outcome with n under ~10 gets a dash, not a percentage (plan §7).")
per_out = data.get("grade_status_by_outcome", {})
display(pd.DataFrame(
    [{"outcome": f"{num} {name}", **{c: per_out.get(num, {}).get(c, 0) for c in COLS}}
     for num, name in ((n, next(o["outcome_name"] for r in data["detailed_records"]
                                for k, o in r["outcomes"].items() if k == n))
                       for n in prov["outcomes"])]))

print("\nSEEDED vs UNSTRATIFIED HOLDOUT - this is the size of the selection bias.")
print("If the holdout looks far worse, the seeds are feeding it the easy cases.")
by_sel = data.get("grade_status_by_selection", {})
if by_sel:
    rows = []
    for k, c in by_sel.items():
        n = sum(c.values())
        rows.append({"selection": k, "pairs": n,
                     **{col: f"{c.get(col, 0)} ({100 * c.get(col, 0) / n:.0f}%)" for col in COLS}})
    display(pd.DataFrame(rows))

print("\nIf most pairs are `cannot_grade`, extraction is too sparse to grade with, however\n"
      "precise the few extractions are. But a high `absent` count is NOT automatically bad -\n"
      "Step 11 is what decides whether those absences are correct.")

feats = data["features_extracted"]
display(pd.DataFrame(sorted(feats.items(), key=lambda kv: -kv[1]),
                     columns=["Feature", "Times accepted"]))

# ---- before/after, when Step 7 saved a previous run ----
PREV_PATH = "results/a100_27b_ollama_prev.json"
if os.path.exists(PREV_PATH):
    with open(PREV_PATH, encoding="utf-8") as f:
        prev = json.load(f)
    p_prov, p_status = prev["provenance"], prev["grade_status"]
    p_total = sum(p_status.values())
    p_recs = prev["detailed_records"]
    print(f"\nBEFORE / AFTER   {p_prov.get('cohort', 'loose')} -> {prov.get('cohort', 'loose')}")
    rows = [("SCD-primary notes",
             f"{sum(1 for r in p_recs if r.get('scd_primary'))}/{len(p_recs)}",
             f"{n_primary}/{len(recs)}")]
    for col in COLS:
        b, a = p_status.get(col, 0), status.get(col, 0)
        if b or a:
            rows.append((col, f"{b} ({100 * b / p_total:.0f}%)", f"{a} ({100 * a / total:.0f}%)"))
    b_dec, a_dec = p_total - p_status.get("absent", 0), total - status.get("absent", 0)
    rows.append(("decided (non-absent)",
                 f"{b_dec} ({100 * b_dec / p_total:.0f}%)", f"{a_dec} ({100 * a_dec / total:.0f}%)"))
    display(pd.DataFrame(rows, columns=["", "before", "after"]))
    print("Note selection changes with the cohort, so these are two different samples -")
    print("read it as 'the measurement got cleaner', not as a per-note improvement.")
```

## Step 11 — Every accepted finding, with its quote

This is the audit surface: feature, value, and the exact span that justified it.

```python
import sys
sys.path.insert(0, "scripts/experiments")

RUN_ID = prov.get("run_id", "")

# Read the harness's OWN record of what it accepted. Re-deriving acceptance here -
# re-running the quote match and the coercion - looks equivalent and is not: any
# drift between the two copies shows up as a review sheet that quietly describes a
# different run than the one it was generated from.
rows, conflict_rows = [], []
for rec in data["detailed_records"]:
    for num, o in rec["outcomes"].items():
        if "accepted_findings" not in o:
            raise SystemExit(
                "This result file predates the per-finding record. Re-run Step 7 with the "
                "current harness; the sheets below cannot be built without it.")
        gr = o["grade_result"]
        for f in o["accepted_findings"]:
            rows.append({
                "run_id": RUN_ID,
                "uid": rec["patient_uid"],
                "outcome": f"{num} {o['outcome_name'][:26]}",
                "feature": f["feature"],
                "value": f["value"],
                "quote": (f["quote"] or "")[:110],
                "unit_note": f.get("unit") or "",
                "status": gr["status"],
                "grade": gr.get("grade"),
            })
        for name, vals in (o.get("conflicts") or {}).items():
            conflict_rows.append({
                "run_id": RUN_ID,
                "uid": rec["patient_uid"],
                "outcome": f"{num} {o['outcome_name'][:26]}",
                "feature": name,
                "candidates": ", ".join(str(v) for v in vals),
                "quotes": " || ".join((f["quote"] or "")[:60]
                                      for f in o["accepted_findings"]
                                      if f["feature"] == name),
            })

df = pd.DataFrame(rows)
conf = pd.DataFrame(conflict_rows)

# The one guarantee worth asserting: this sheet counts exactly what the run counted.
expected = data["runs"][0]["accepted"]
assert len(df) == expected, f"sheet has {len(df)} rows, run accepted {expected}"

print(f"{len(df)} accepted findings across {df['uid'].nunique()} notes"
      if len(df) else "No accepted findings.")
pd.set_option("display.max_colwidth", 115)
display(df)

if len(conf):
    print(f"\n{len(conf)} feature(s) withheld from grading: several verified values, and no")
    print("aggregation rule in the schema to choose between them. A note carries five")
    print("creatinines across twelve years and one of them belongs to the transplant donor;")
    print("there is no rule that says which is 'the' creatinine, so none was invented.")
    print("Read these first - they are where the extractor and the note disagree most.")
    display(conf)
```

## Step 12 — Hand-check worksheet (the number that decides Arm B) ⭐

Quote-verification says the span is real. It does **not** say the span supports the value —
`vasopressors = true` quoting *"aggressive fluid resuscitation with lactated ringers along with
phenylephrine bolus and drip"* is verbatim, and judging it takes a human.

`tasks/medgemma_extraction_test.md` gates the decision on **precision**:

| Precision | Read |
|:--|:--|
| ≥ 90% | Arm B viable |
| 75–90% | good as a third engine, not alone |
| < 75% | a labeller, not an extractor — revisit the prompt before the model |

Fill in `supports_value` with `y` / `n`. Budget ~15–20 s per row, ~1.5 h total. At n=100 the 95%
interval is roughly ±6pp near 90%, so read the result as a band, not a point.

```python
HANDCHECK_PATH = f"results/handcheck_stage{REVIEW_STAGE}.csv"
CONFLICT_PATH = f"results/conflicts_stage{REVIEW_STAGE}.csv"

if len(df):
    work = df.copy()
    work.insert(len(work.columns), "supports_value", "")        # <- you fill this in: y / n
    work.insert(len(work.columns), "reviewer_note", "")
    work = work.sample(n=min(100, len(work)), random_state=0)   # capped per the P11 protocol
    work.to_csv(HANDCHECK_PATH, index=False)
    print(f"Wrote {len(work)} rows to {HANDCHECK_PATH}   (run_id {RUN_ID})")
    print("\nEvery row already passed quote verification. That check asks whether the words")
    print("are in the note; it cannot ask whether they support the value, and rows that fail")
    print("that second question pass the first one cleanly. This column is that question.")
    print("\nOpen it, mark supports_value as y/n, then compute:")
    print("   precision = y / (y + n)")
    print("\nAlso read 5 full notes and list what SHOULD have been extracted for the four")
    print("outcomes, then diff against the rows above. That is the recall check - directional")
    print("at n=5, which is fine: the question is whether it misses things plainly stated.")
else:
    print("Nothing accepted to hand-check. That is itself the finding - extraction is too sparse.")

if len(conf):
    conf.to_csv(CONFLICT_PATH, index=False)
    print(f"\nWrote {len(conf)} withheld-conflict rows to {CONFLICT_PATH}")
```

## Step 13 — Absence audit (does "none" mean none?) ⭐

Step 10 measures the **present** side: of what the model extracted, how much is right. It says
nothing about what it *missed*.

Almost every `absent` verdict comes straight from the model's `present: false` flag, and nothing
in the pipeline checks it. That flag decides whether an outcome is graded at all — so it gates
everything Step 10 measures, and it is currently the least-examined decision in the system.

`tasks/plan.md` §7 names this:

**Two sheets come out of this step, because there are two questions.** `absent` means the
model never evidenced the outcome — that is the false-negative question below. `refuted` means
the model *did* evidence it and the decision tables overruled the call (a 36.5 °C "fever").
Confirming a refuted pair re-checks the rule engine, not the model, so it goes to its own
sheet and stays out of the false-negative denominator.

> **Absence audit.** Human-confirm a random sample of (note, outcome) pairs marked *absent* →
> false-negative rate estimate. This is the number that protects the absent / cannot-determine
> distinction (§1).

Read the note, then answer one question: **is this outcome genuinely not in this note?**
`y` = correctly absent. `n` = a false negative, the outcome is there and was missed.

Sort by `uid` so you read each note once and answer for all four outcomes together — about
10 minutes per note. A false-negative rate here is the counterweight to precision: a model can
score 100% precision by extracting almost nothing, and only this catches that.

```python
from medgemma_extraction import feature_brief, prompt_features, stage

ABSENCE_PATH = f"results/absence_audit_stage{REVIEW_STAGE}.csv"
REFUTED_PATH = f"results/refuted_audit_stage{REVIEW_STAGE}.csv"

_ST = stage(prov.get("prompt_stage", "0"))

def what_would_make_it_present(num):
    # The exact feature list the model was shown, so the reviewer checks the same
    # thing. Read from the same function the prompt is built from rather than by
    # splitting the rendered text: the headings move between rungs, and a split that
    # silently returns nothing gives the reviewer an empty column to audit against.
    return " | ".join(feature_brief(n, num, _ST) for n in prompt_features(num, _ST))

# Two different questions, and they were previously one pile:
#   absent  - the model never evidenced the outcome. Was it there? -> false-negative rate.
#   refuted - the model DID evidence it and the tables overruled the call (a 36.5 degC
#             "fever"). Asking a reviewer to confirm THAT absence asks them to re-check
#             the rule engine, not the model, and pollutes the false-negative estimate.
abs_rows, ref_rows = [], []
for rec in data["detailed_records"]:
    for num, o in rec["outcomes"].items():
        gr = o["grade_result"]
        if gr["status"] not in ("absent", "refuted"):
            continue
        row = {
            "run_id": RUN_ID,
            "uid": rec["patient_uid"],
            "selection": rec.get("selection", ""),
            "outcome": num,
            "outcome_name": o["outcome_name"],
            "model_said_present": o.get("present"),
            "truly_absent": "",                      # <- you fill this in: y / n
            "reviewer_note": "",
            "look_for": what_would_make_it_present(num),
            "title": rec.get("title", ""),
            "note_text": rec["patient_note"],
        }
        if gr["status"] == "absent":
            abs_rows.append(row)
        else:
            row["extracted_features"] = json.dumps(o["extracted_features"])
            row["rule_reason"] = gr.get("reason") or ""
            ref_rows.append(row)

adf = pd.DataFrame(abs_rows)
if len(adf):
    sample = adf.sample(n=min(50, len(adf)), random_state=0).sort_values(["uid", "outcome"])
    sample.to_csv(ABSENCE_PATH, index=False)
    print(f"{len(adf)} pairs the MODEL marked absent; wrote {len(sample)} to {ABSENCE_PATH}")
    print(f"   (run_id {RUN_ID} - stamped on every row, so a filled sheet can always be")
    print( "    matched back to the run it came from)")
    print("\n  false-negative rate = n / (y + n)")
    print("  A high rate means the model is silently skipping outcomes that are present,")
    print("  which no precision number will ever reveal.")
    display(sample.groupby("outcome_name").size().rename("absent pairs").to_frame())
else:
    print("Nothing marked absent - unusual, and worth checking before trusting the run.")

rdf = pd.DataFrame(ref_rows)
if len(rdf):
    rdf.to_csv(REFUTED_PATH, index=False)
    print(f"\n{len(rdf)} pairs where the model said present and the TABLES said no.")
    print(f"Wrote them to {REFUTED_PATH} - a separate question, and a much cheaper one:")
    print("these are over-calls the deterministic layer already caught, so they cost")
    print("nothing downstream. They belong in the absence audit only if the rule was wrong.")
    display(rdf[["uid", "outcome", "outcome_name", "extracted_features", "rule_reason"]])
```
